```python
!pip install sdv
import sdv

print(sdv.version.public)
```

```python
from sdv.metadata import SingleTableMetadata
from sdv.single_table import CTGANSynthesizer

# Step 1: Define metadata
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(female_df_1)

# Step 2: Initialize synthesizer with metadata
synthesizer = CTGANSynthesizer(metadata)

# Step 3: Fit the model
synthesizer.fit(female_df_1)

# Step 4: Generate synthetic data
synthetic_female_df_1 = synthesizer.sample(num_rows=84)

# Step 5: Preview
display(synthetic_female_df_1.head())

augmented_female_df_1 = pd.concat([female_df_1, synthetic_female_df_1], ignore_index=True)

print(f"Total rows after augmentation: {len(augmented_female_df_1)}")
```

```python
female_df_0 = female_df[female_df['target']==0]
augmented_female_df = pd.concat([augmented_female_df_1,female_df_0],ignore_index=True)

```

```python
augmented_female_df['target'].value_counts()
```

```python
balanced_df = pd.concat([augmented_female_df, male_df],ignore_index=True)
from sklearn.utils import shuffle
balanced_df = shuffle(balanced_df)
```

```python
from sdv.metadata import SingleTableMetadata
from sdv.single_table import CTGANSynthesizer

# Step 1: Define metadata
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(female_df_1)

# Step 2: Initialize synthesizer with metadata
synthesizer = CTGANSynthesizer(metadata)

# Step 3: Fit the model
synthesizer.fit(female_df_1)

# Step 4: Generate synthetic data
synthetic_female_df_1 = synthesizer.sample(num_rows=324)

# Step 5: Preview
display(synthetic_female_df_1.head())

augmented_female_df_1 = pd.concat([female_df_1, synthetic_female_df_1], ignore_index=True)

print(f"Total rows after augmentation: {len(augmented_female_df_1)}")
```

```python
remerged_female_df = pd.concat([augmented_female_df_1,augmented_female_df_0],ignore_index=True)
remerged_female_df['target'].value_counts()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split

X = rebalanced_df.drop('target',axis=1)
y = rebalanced_df['target']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,stratify=y, random_state=123)

model = RandomForestClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```

# Bias mitigation for clinical

```python
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/sample_data/combined_statslogExcluded_SMOTE_GenderBalancedWithCTGAN.csv')
X = df.drop('target',axis=1)
y = df['target']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=123,stratify=y)
df.shape
```

```python
!pip install fairlearn
from fairlearn.metrics import (MetricFrame,
    demographic_parity_difference,
    demographic_parity_ratio,
    equalized_odds_difference,
    equalized_odds_ratio, selection_rate
)
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
import xgboost as xgb
from sklearn.metrics import classification_report
xgb_clf = xgb.XGBClassifier(
    objective="binary:logistic",  # or "multi:softprob" for multiclass
    eval_metric="logloss",        # can be "mlogloss" or "auc" based on task
    use_label_encoder=False,
    n_estimators = 115, max_depth = 6, learning_rate = 0.0547, subsample = 0.5715,
    colsample_bytree = 0.5422, gamma = 1.5894, reg_alpha = 0.9968, reg_lambda = 0.8258
)

xgb_clf.fit(X_train, y_train)
y_pred = xgb_clf.predict(X_test)
print(classification_report(y_pred, y_test))

print("Bias metrics shown below:")
print("--"*40)
print("--"*40)

# Bias metrics check
y_true = y_test.copy()
sensitive_features = X_test['sex']  # 0: Female, 1: Male

# ➤ Create the fairness metrics frame
mf = MetricFrame(
    metrics={
        'accuracy': accuracy_score,
        'selection_rate': selection_rate,
        'precision': precision_score,
        'recall': recall_score
    },
    y_true=y_true,
    y_pred=y_pred,
    sensitive_features=sensitive_features
)

print("Group-wise metrics:")
print(mf.by_group)

# ➤ Fairness differences and ratios
dp_diff = demographic_parity_difference(y_true, y_pred, sensitive_features=sensitive_features)
dp_ratio = demographic_parity_ratio(y_true, y_pred, sensitive_features=sensitive_features)

eo_diff = equalized_odds_difference(y_true, y_pred, sensitive_features=sensitive_features)
eo_ratio = equalized_odds_ratio(y_true, y_pred, sensitive_features=sensitive_features)

print("\nDemographic parity difference:", dp_diff)
print("Demographic parity ratio:", dp_ratio)

print("\nEqualized odds difference:", eo_diff)
print("Equalized odds ratio:", eo_ratio)

```
