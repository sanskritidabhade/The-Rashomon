```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Drop non-numeric columns (e.g., 'target' if it's categorical or ID fields)
df_numeric = df.select_dtypes(include=['float64', 'int64'])

# Optional: Drop 'target' column for clustering if it's the label
if 'target' in df_numeric.columns:
    df_numeric = df_numeric.drop(columns=['target'])

# Fill missing values with column-wise median
df_filled = df_numeric.fillna(df_numeric.median())

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_filled)

# Reduce dimensionality for better clustering performance
pca = PCA(n_components=6, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Apply KMeans clustering to split into 4 regions
kmeans = KMeans(n_clusters=4, random_state=42, n_init='auto')
cluster_labels = kmeans.fit_predict(X_pca)

# Evaluate cluster quality
silhouette = silhouette_score(X_pca, cluster_labels)
print(f"Silhouette Score (clustering quality): {silhouette:.4f}")

# Append cluster labels to the original dataset
df['Cluster_region'] = cluster_labels

# Save the labeled dataset
#df.to_csv("combined_with_theoretical_region.csv", index=False)
#print("✅ Clustered dataset saved as: combined_with_theoretical_region.csv")
df['Cluster_region'].value_counts()
```
