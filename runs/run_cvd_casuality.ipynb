```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score as acc, mean_absolute_error as mae
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/sample_data/combined_statslogExcluded_SMOTE_GenderBalancedWithCTGAN.csv')
support_df = df.drop('target',axis=1)
```

```python
print(support_df.columns.to_list())
```

# causal inference

```python
treatment = 'age'

X = support_df.drop(treatment,axis=1)
y = support_df[treatment]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123)

rf_model = RandomForestRegressor()
rf_model.fit(X_train, y_train)
# Assume rf_model is your fitted RandomForestClassifier
importances = rf_model.feature_importances_
feat_names = X_train.columns
feat_df = pd.DataFrame({'feature': feat_names, 'importance': importances})
feat_df = feat_df.sort_values(by='importance', ascending=False)
feat_df['cumulative_importance'] = feat_df['importance'].cumsum()
# Select features that together explain 80% of total importance
top_feats = feat_df[feat_df['cumulative_importance'] <= 0.80]['feature'].tolist()
# Print treatment and confounders
print('Treatment:',treatment)
print("Selected confounders based on RF:", top_feats)

# Define causal model
model = CausalModel(
    data=df,  # Full dataset
    treatment= treatment,  # The feature whose causal effect you want to measure
    outcome="target",  # The target variable
    common_causes=top_feats  # The identified confounders
)
# Identify the causal effect
identified_estimand = model.identify_effect()

# Doubly robust estimation
effect_dr = model.estimate_effect(identified_estimand, method_name="backdoor.linear_regression")

# Print causal effect
print(effect_dr)
```

# Hybrid Modeling
Hybrid model will be trained on treatment/feature with |ATE| >= 1
These features are hence: ['exang', 'slope', 'fbs', 'cp', 'sex', 'oldpeak']


```python
reduced_df = df[['exang', 'slope', 'fbs', 'cp', 'sex', 'oldpeak']]
reduced_df
```

```python
reduced_df = df[['exang', 'slope', 'fbs', 'cp', 'sex', 'oldpeak','target']]
X = reduced_df.drop('target',axis=1)
y = reduced_df['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, stratify=y, random_state=123)
```

```python
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score, precision_score, classification_report
```

```python


# Define base models
voting_models = [
    ('xgb', XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)),

    ('lgbm', LGBMClassifier(random_state=42)),

    ('rf', RandomForestClassifier(random_state=42)),
    ('logreg', LogisticRegression(max_iter=1000)),
    ('knn', KNeighborsClassifier(n_neighbors=5))
]

# Build Voting Classifier (soft voting)
voting_clf = VotingClassifier(estimators=voting_models, voting='soft')
voting_clf.fit(X_train, y_train)

# Predict
voting_preds = voting_clf.predict(X_test)

# Evaluate
accuracy = accuracy_score(y_test, voting_preds)
precision = precision_score(y_test, voting_preds)
report = classification_report(y_test, voting_preds)

print("🔍 Voting Classifier Accuracy:", round(accuracy, 4))
print("🎯 Precision:", round(precision, 4))
print("📊 Classification Report:\n", report)
```

```python
import pandas as pd
proba = voting_clf.predict_proba(X_test)[:, 1]
out = pd.DataFrame({
    "row_id": X_test.index,
    "y_true": y_test.values,
    "y_pred": voting_preds,
    "proba_1": proba,
    "model": "CGEVC_author_params",
    "protocol": "author_split_rs123_leaky",
})
out.to_csv("preds_cgevc.csv", index=False)
print(out.shape, (out.y_pred == out.y_true).mean())
```
